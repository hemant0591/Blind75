### 21. Merge Two Sorted Lists

You are given the heads of two sorted linked lists list1 and list2.

Merge the two lists into one sorted list. The list should be made by splicing together the nodes of the first two lists.

Return the head of the merged linked list.

 

Example 1:


Input: list1 = [1,2,4], list2 = [1,3,4]
Output: [1,1,2,3,4,4]
Example 2:

Input: list1 = [], list2 = []
Output: []
Example 3:

Input: list1 = [], list2 = [0]
Output: [0]

```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        # define an empty node
        dummy = ListNode()
        # set the variable to dummy to move it along the chain
        tail = dummy

        # run the loop while both lists exist
        while list1 and list2:
            # check which node has lower value
            if list1.val < list2.val:
                tail.next = list1
                # move list1
                list1 = list1.next
            else:
                tail.next = list2
                list2 = list2.next
            # move tail forward
            tail = tail.next
        
        # check for the remainder list
        tail.next = list1 if list1 else list2

        # return the head stored in dummy.next
        return dummy.next
```
