### 40. Combination Sum II

Given a collection of candidate numbers (candidates) and a target number (target), find all unique combinations in candidates where the candidate numbers sum to target.

Each number in candidates may only be used once in the combination.

Note: The solution set must not contain duplicate combinations.

 

Example 1:

Input: candidates = [10,1,2,7,6,1,5], target = 8
Output: 
[
[1,1,6],
[1,2,5],
[1,7],
[2,6]
]
Example 2:

Input: candidates = [2,5,2,1,2], target = 5
Output: 
[
[1,2,2],
[5]
]

```python
class Solution:
    def combinationSum2(self, candidates: List[int], target: int) -> List[List[int]]:
        candidates.sort()
        res = [] # stores all unique combinations
        path = [] # stores current path taken

        def dfs(index, rem):
            """
            index: current index being processed
            rem: remaining value from target - sum(combinations)
            """

            # base cases
            if rem == 0:
                # we found a solution, add to res
                res.append(path.copy())
                return
            if rem < 0:
                # we've overshot, return from current path
                return

            # process the indices
            for i in range(index, len(candidates)):
                # check if we're using the same element twice
                if i > index and candidates[i] == candidates[i - 1]:
                    continue

                c = candidates[i]
                # append candidate to 
                path.append(c)
                # make recursive call using next index (not allowed to use the same element twice)
                dfs(i + 1, rem - c)
                # remove from path after exploring
                path.pop()

        dfs(0, target)
        return res
```
